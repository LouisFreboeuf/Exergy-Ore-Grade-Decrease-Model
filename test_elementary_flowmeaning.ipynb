```python
import bw2data as bd
bd.projects.set_current("CERC")
print(bd.databases)  # List available databases
def inspect_biosphere_flow(flow_name, flow_category, biosphere_db_name, technosphere_db_name):
    # Load the databases as objects
    biosphere_db = bd.Database(biosphere_db_name)
    technosphere_db = bd.Database(technosphere_db_name)

    # Find the biosphere flow by name & category
    matching_flows = [
        flow for flow in biosphere_db # type: ignore
        if flow.get("name", "").lower() == flow_name.lower()
        and flow.get("categories", ()) == flow_category
    ]

    if not matching_flows:
        print("No matching flows found.")
        return

    target_flow = matching_flows[0]
    print(f"\n=== Flow found: {target_flow['name']} ({target_flow.get('unit')}) ===")
    print(f"Key: {target_flow.key}")

    # Now search for an exchange that uses this flow
    for act in technosphere_db: # type: ignore
        for exc in act.exchanges():
            if exc.input == target_flow.key:  # match by flow key
                print(f"\nFound in activity: {act['name']}")
                props = exc.get("properties", {})
                if not props:
                    print("  No properties attached to this exchange.")
                else:
                    for pname, pdata in props.items():
                        print(f"  {pname}: {pdata.get('amount')} {pdata.get('unit')}")
                cas_number = exc.get("CAS number", {})
                if not cas_number:
                    print("  No CAS number attached to this exchange.") 
                else:
                    print(f"  CAS number: {cas_number}")
                synonyms = exc.get("synonyms", {})
                if not synonyms:
                    print("  No synonyms attached to this exchange.")   
                else:
                    print(f"  Synonyms: {synonyms}")
                return

    print("No exchange found with this flow.")

# Example call: Replace with your actual database names
inspect_biosphere_flow(
    "Silver, 3.2ppm in sulfide, Ag 1.2ppm, Cu and Te, in crude ore, in ground",
    ("natural resource", "in ground"),
    list(bd.databases)[0],  # Replace with your biosphere database name
    list(bd.databases)[1]  # Replace with your technosphere database name
)
```
